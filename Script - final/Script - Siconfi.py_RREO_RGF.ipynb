```python
# df_RGF = get_fiscal(year = 2025, period = [1, 2, 3], cod = 27, power = ['E', 'L','J'])

ANOS = [2022, 2023, 2024, 2025, 2026]
PERIODOS = [1, 2, 3]
```

```python

DF_AL_RGF = get_fiscal(year=ANOS, period=PERIODOS, cod=27, power=["E"])
DF_AL_RGF1 = get_fiscal(year=ANOS, period=PERIODOS, cod=27, power=["L"])
DF_AL_RGF2 = get_fiscal(year=ANOS, period=PERIODOS, cod=27, power=["J"])


# 21 – Maranhão (MA)
DF_MA_RGF  = get_fiscal(year=ANOS, period=PERIODOS, cod=21, power=["E"])
DF_MA_RGF1 = get_fiscal(year=ANOS, period=PERIODOS, cod=21, power=["L"])
DF_MA_RGF2 = get_fiscal(year=ANOS, period=PERIODOS, cod=21, power=["J"])

# 22 – Piauí (PI)
DF_PI_RGF  = get_fiscal(year=ANOS, period=PERIODOS, cod=22, power=["E"])
DF_PI_RGF1 = get_fiscal(year=ANOS, period=PERIODOS, cod=22, power=["L"])
DF_PI_RGF2 = get_fiscal(year=ANOS, period=PERIODOS, cod=22, power=["J"])

# 23 – Ceará (CE)
DF_CE_RGF  = get_fiscal(year=ANOS, period=PERIODOS, cod=23, power=["E"])
DF_CE_RGF1 = get_fiscal(year=ANOS, period=PERIODOS, cod=23, power=["L"])
DF_CE_RGF2 = get_fiscal(year=ANOS, period=PERIODOS, cod=23, power=["J"])

# 24 – Rio Grande do Norte (RN)
DF_RN_RGF  = get_fiscal(year=ANOS, period=PERIODOS, cod=24, power=["E"])
DF_RN_RGF1 = get_fiscal(year=ANOS, period=PERIODOS, cod=24, power=["L"])
DF_RN_RGF2 = get_fiscal(year=ANOS, period=PERIODOS, cod=24, power=["J"])

# 25 – Paraíba (PB)
DF_PB_RGF  = get_fiscal(year=ANOS, period=PERIODOS, cod=25, power=["E"])
DF_PB_RGF1 = get_fiscal(year=ANOS, period=PERIODOS, cod=25, power=["L"])
DF_PB_RGF2 = get_fiscal(year=ANOS, period=PERIODOS, cod=25, power=["J"])

# 26 – Pernambuco (PE)
DF_PE_RGF  = get_fiscal(year=ANOS, period=PERIODOS, cod=26, power=["E"])
DF_PE_RGF1 = get_fiscal(year=ANOS, period=PERIODOS, cod=26, power=["L"])
DF_PE_RGF2 = get_fiscal(year=ANOS, period=PERIODOS, cod=26, power=["J"])

# 28 – Sergipe (SE)
DF_SE_RGF  = get_fiscal(year=ANOS, period=PERIODOS, cod=28, power=["E"])
DF_SE_RGF1 = get_fiscal(year=ANOS, period=PERIODOS, cod=28, power=["L"])
DF_SE_RGF2 = get_fiscal(year=ANOS, period=PERIODOS, cod=28, power=["J"])

# 29 – Bahia (BA)
DF_BA_RGF  = get_fiscal(year=ANOS, period=PERIODOS, cod=29, power=["E"])
DF_BA_RGF1 = get_fiscal(year=ANOS, period=PERIODOS, cod=29, power=["L"])
DF_BA_RGF2 = get_fiscal(year=ANOS, period=PERIODOS, cod=29, power=["J"])
```

```python
df_RGF_total = pd.concat(
                            [DF_AL_RGF, 
                            DF_AL_RGF1, 
                            DF_AL_RGF2,
                            DF_MA_RGF,
                            DF_MA_RGF1,
                            DF_MA_RGF2,
                            DF_PI_RGF,
                            DF_PI_RGF1,
                            DF_PI_RGF2,
                            DF_CE_RGF,
                            DF_CE_RGF1,
                            DF_CE_RGF2,
                            DF_RN_RGF,
                            DF_RN_RGF1,
                            DF_RN_RGF2,
                            DF_PB_RGF,
                            DF_PB_RGF1,
                            DF_PB_RGF2,
                            DF_PE_RGF,
                            DF_PE_RGF1,
                            DF_PE_RGF2,
                            DF_SE_RGF,
                            DF_SE_RGF1,
                            DF_SE_RGF2,
                            DF_BA_RGF,
                            DF_BA_RGF1,
                            DF_BA_RGF2], 
                          axis=0, 
                          ignore_index=True)
df_RGF_total
```
