Filtrando informações específicas por dataframe

---> Basicamente é aplicação de filtros


----> .isin([...] é para selecionar vários textos

---> aqui tem prestar atenção entre (), [] e {}
- () - serve para funções (sempre tendo protagonismo para os argumentos) e listas/tuplas
- [] - serve para pegar itens específicos da indexação, ex: posição 0 == [0]; condicionados ao acesso, usado principalmente em filtros e colunas
- {} - Serve para guardar informações, ex: {'nome': 'Ana', 'idade': 30} // chave: valor; tem grande serventia em f-strings --> ex: f"O resultado é {df_filtrado}"



```python
df_filtrado = df_oficial[df_oficial['Descrição'].isin(['DESPESAS EMPENHADAS',
    'DESPESAS EMPENHADAS NO BIMESTRE',
    'LIQUIDADOS',
    'Saldo a Pagar',
    'SALDO ATUAL',
    'Saldo a Realizar'
])]

df_filtrado
```
